## Import ##

```python
# LOCATION : https://github.com/purnasai/Dino_V2
import torch
from torchvision import models, transforms
import cv2
import os
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from PIL import Image
import pickle

os.environ["XFORMERS_DISABLED"] = "1" # Switch to enable xFormers
device = torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")
```

## Evaluation ##

```python
## Take right hand average 
frame_frequency = 5
sample_size = 275
right_hand_pickle_file = open('/media/osero/SamsungSSD/pickles/features_right_hand_frames_small_ordered.pickle', 'rb')
right_hand_paths, right_hand_features, right_hand_labels = pickle.load(right_hand_pickle_file)
right_hand_average_features = []

for feature in right_hand_features:
    sampled_features = feature[0::frame_frequency]
    right_hand_average_features.append(np.mean(sampled_features, axis=0))
    ccc = 5

unique_values = []
limitter_counter = 0
# Loop through the original list
for item in right_hand_labels:
    if item not in unique_values:
        unique_values.append(item)
    if(len(unique_values) > sample_size):
        break
    limitter_counter += 1
print(limitter_counter)
        
right_hand_average_features = right_hand_average_features[0:limitter_counter]
right_hand_labels = right_hand_labels[0:limitter_counter]
ccc = 5
```

```python
## Take left hand average 

left_hand_pickle_file = open('/media/osero/SamsungSSD/pickles/features_left_hand_frames_small_ordered.pickle', 'rb')
left_hand_paths, left_hand_features,left_hand_labels = pickle.load(left_hand_pickle_file)
left_hand_average_features = []

for feature in left_hand_features:
    sampled_features = feature[0::frame_frequency]
    left_hand_average_features.append(np.mean(sampled_features, axis=0))
    ccc = 5

left_hand_average_features = left_hand_average_features[0:limitter_counter]
left_hand_labels = left_hand_labels[0:limitter_counter]
ccc = 5
```

```python
average_features = []
labels = left_hand_labels
for i in range (0,len(left_hand_labels)):
    average_features.append(np.concatenate((left_hand_average_features[i], right_hand_average_features[i],)))
cc = 5
print('Label Match Result: ', left_hand_labels == right_hand_labels)
print('Paths Match Result: ', [text.split('/')[-2:] for text in left_hand_paths] == [text.split('/')[-2:] for text in right_hand_paths])
```

```python
num_of_test = 20
accuracy_list = []
for i in range(0,num_of_test):
    # Step 5: Train a Classifier on the Video Embeddings
    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(average_features, labels, test_size=0.2)

    # Initialize a k-NN classifier
    knn = KNeighborsClassifier(n_neighbors = 1)

    # Train the classifier
    knn.fit(X_train, y_train)

    # Predict on the test set and calculate accuracy
    y_pred = knn.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracy_list.append(accuracy)
    print(f"{i}th accuracy for Right+Left: {accuracy * 100:.2f}%")
average_accuracy = np.mean(accuracy_list)
print(f"Average Accuracy for Right+Left: {average_accuracy * 100:.2f}%")
```
