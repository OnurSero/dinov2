```python
import cv2 
from matplotlib import pyplot as plt
from PIL import Image
import pickle
import shutil
import os
import glob
import csv
import numpy as np
from collections import OrderedDict
```

```python
import contextlib
import gc

@contextlib.contextmanager
def clear_memory():
    try:
        yield
    finally:
        gc.collect()
```

```python
main_pickle_folder = "/media/osero/SamsungSSD/pickles" 
main_feature_folder = "/media/osero/SamsungSSD/features" 
found = False


for pickle_file_name in sorted(os.listdir(main_pickle_folder)):
    gc.collect()
    with clear_memory():   
        feature_name = pickle_file_name.removesuffix(".pickle")
        pickle_path = os.path.join(main_pickle_folder, pickle_file_name)

        if 'features_left_hand_frames_large_train' in feature_name and not found:
            found = True
            continue

        if not os.path.isfile(pickle_path):
            continue
        if 'bsign22_heatmap' in pickle_path:
            continue
        if not found:
            continue
        
        print(pickle_path)
        pickle_file = open(pickle_path, 'rb')
        paths, features, labels = pickle.load(pickle_file)
        for i, feature_path in enumerate(paths):
            filename = os.path.splitext(os.path.basename(feature_path))[0]
            parent_folder = os.path.basename(os.path.dirname(feature_path))
            result_pickle_path = os.path.join(main_feature_folder, feature_name, parent_folder, filename, "feature.pickle")
            os.makedirs(os.path.dirname(result_pickle_path), exist_ok=True)
            with open(result_pickle_path, 'wb') as handle:
                pickle.dump(features[i], handle, protocol=pickle.HIGHEST_PROTOCOL)

        
```
